```python
import numpy as np
```

```python
#PROBLEM A (REPRODUCIBLE NORMALIZATION)
np.random.seed(211)
X = np.random.randint(10, 101, size=(5, 5))
```

```python
x_bar = np.mean(X)
sigma = np.std(X)
X_normalized = (X - x_bar) / sigma
```

```python
print( "PROBLEM A" )
```

```python
print("Original Array (X):\n", X)
```

```python
print("\nNormalized Array (X_normalized):\n", X_normalized)
```

```python
print("\nMean of Normalized Array:", np.mean(X_normalized))
```

```python
print("Std Dev of Normalized Array:", np.std(X_normalized))
```

```python
np.save("X_normalized.npy", X_normalized)
```

```python
#PROBLEM B(CUBES DIVISIBLE BY 4)
C = (np.arange(1, 101) ** 3).reshape(10, 10)
div_by_4 = C[C % 4 == 0]
```

```python
print("Problem B")
```

```python
print("Shape of C:", C.shape)
```

```python
print("Elements Divisible by 4:\n", div_by_4)
```

```python
print("Number of Selected Elements:", len(div_by_4))
```

```python
np.save("div_by_4.npy", div_by_4)
```

```python
#PROBLEM C(ABOVE-MEAN SQUARES)
S = (np.arange(1, 37) ** 2).reshape(6, 6)
S_mean = np.mean(S)
```

```python
above_mean = S[S>S_mean]
```

```python
print("Problem C")
```

```python
print("Array(S):\n", S)
```

```python
print("Mean of S:", S_mean)
```

```python
print("Elements Above Mean:\n", above_mean)
```

```python
print("Number of Selected Elements:", len(above_mean))
```

```python
np.save("above_mean.npy", above_mean)
```
